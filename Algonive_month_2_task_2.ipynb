```python
# Task 5:- Engineering New Features
```

```python
# Convert 'last_review' to datetime, coercing errors (including 'No reviews') to NaT
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')

# Extract the year from 'last_review'
df['review_year'] = df['last_review'].dt.year

# Drop rows where 'review_year' is NaN (which correspond to 'No reviews' or invalid dates)
df_with_reviews = df.dropna(subset=['review_year'])

# Get the count of bookings per year
year_bookings = df_with_reviews['review_year'].value_counts().sort_index()
display(year_bookings)
```

```python
# Find the year with the highest number of bookings
top_booking_year = year_bookings.idxmax()
top_booking_count = year_bookings.max()

plt.figure(figsize=(10,5))
sns.barplot(x=year_bookings.index, y=year_bookings.values, palette='viridis')
plt.title(f'Year with the Highest Number of Bookings: {top_booking_year} with {top_booking_count} bookings')
plt.xlabel('Year')
plt.ylabel('Number of Bookings')
plt.show()
```
