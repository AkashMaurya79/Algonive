```python
# Task 3:- Exploratory Data Analysis (EDA)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
plt.figure(figsize=(10,5))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True)
plt.show()
```

```python
df.Gender.value_counts()
```

```python
plt.figure(figsize=(10,5))
gender_counts = df['Gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Gender')
plt.show()
```

```python
df['Product Category'].value_counts()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Product Category', data=df, palette='viridis')
plt.title('Distribution of Product Categories')
plt.show()
```

```python
# Observation of Task 3:-
# In this task , I Observe the gender contribution of that dataset that is 490 are male and 510 are female & in the percentage ratio the male are 49% and 51% are female.
# And also I observe the total products according to category which is purchase by the customer that is clothing-351,Electronics-342,Beauty-307.
```

```python
# Task 4:- Data Visualization
```

```python
age_with_gender = df.groupby(['Age', 'Gender']).size().reset_index(name='Count')
age_with_gender
```

```python
plt.figure(figsize=(15,5))
sns.barplot(x='Age', y='Count', hue='Gender', data=age_with_gender, palette='viridis')
plt.title('Age Distribution by Gender')
plt.show()
```

```python
gender_counts_catgory=df.groupby(['Gender','Product Category']).size().reset_index(name='Count')
gender_counts_catgory
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Product Category', hue='Gender', data=df, palette='viridis')
plt.title('Distribution of Product Categories by Gender')
plt.show()
```

```python
product_Quantity=df.groupby(['Product Category','Quantity']).size().reset_index(name='Count')
product_Quantity
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category', y='Count', hue='Quantity', data=product_Quantity, palette='viridis')
plt.title('Product Category Distribution by Quantity')
plt.show()
```

```python
Quantity_with_gender = df.groupby(['Quantity', 'Gender']).size().reset_index(name='Count')
Quantity_with_gender
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Quantity', y='Count', hue='Gender', data=Quantity_with_gender, palette='viridis')
plt.title('Quantity Distribution by Gender')
plt.show()
```

```python
plt.figure(figsize=(10,5))
labels = gender_counts_catgory['Gender'] + ' - ' + gender_counts_catgory['Product Category']
plt.pie(gender_counts_catgory['Count'], labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Product Categories by Gender')
plt.show()
```
