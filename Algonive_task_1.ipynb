```python
# Task 3:- Exploratory Data Analysis (EDA)
```

```python
df['Country'].nunique()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Country',data=df)
plt.xticks(rotation=90)
plt.title('Contribution of Country')
plt.show()
```

```python
top_Unit_Price=df.groupby('UnitPrice')['Quantity'].sum().sort_values(ascending=False).head(10)
top_Unit_Price
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top_Unit_Price.index,y=top_Unit_Price.values,palette='viridis')
plt.xticks(rotation=90)
plt.ylabel('Quantity')
plt.title('Top 10 Unit Price according to Quantity')
plt.show()
```

```python
# Observation of Task 3:-
# In this task , I Observe the country contribution and I found that there is 38 countries that part of the dataset but the United Kingdom have maximum contribution.
# And also I observe the top 10 unit price according to Quantity and I found that 1.25 have maximum Quantity.
```

```python
# Task 4:- Data Visualization
```

```python
top_country=df.groupby('Country')['Quantity'].sum().sort_values(ascending=False).head(10)
top_country
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top_country.index,y=top_country.values)
plt.xticks(rotation=90)
plt.ylabel('Quantity')
plt.title('Top 10 Country with respect Quantity')
plt.show()
```

```python
top_customer=df.groupby('CustomerID')['Quantity'].sum().sort_values(ascending=False).head(10)
top_customer
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top_customer.index,y=top_customer.values,palette='viridis')
plt.xticks(rotation=90)
plt.ylabel('Quantity')
plt.title('Top 10 Customer with respect to Quantity')
plt.show()
```

```python
top_stock=df.groupby('StockCode')['Quantity'].sum().sort_values(ascending=False).head(10)
top_stock
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top_stock.index,y=top_stock.values,palette='viridis')
plt.ylabel('Quantity')
plt.title('Top 10 Stock with respect to Quantity')
plt.show()
```

```python
# Observation of Task 4:-
# I observe the Country contribution with respect their Quantity and found that there is United Kingdom have maximum Quantity and the other hand Netherlands have 2nd Number of Quantity.
# And also observe the customer ID with respect the Quantity and found that the null customer that have replaced by zero that have maximum Quantity and the 14646 customer ID have 2nd maximum Quantity.
# And the other hands I observe the top 10 stock code and I found that 22197(Stock Code) have maximum Quantity.
```
